```python
import pandas as pd
schedule = pd.DataFrame(columns= ['content'])
schedule.index.name= 'date'
content = "learn agent"
date = "2026-12-30 8:00" 
date= pd.Timestamp(date)
from datetime import datetime
now = datetime.now()
def w_pd(date: datetime , content: str):
    ''' 
    add content to date index
    '''
    global schedule
    if date not in schedule.index:
        schedule.loc[date, 'content']= ""
    schedule.loc[date,'content'] += content
w_pd(date, content)
```

```python
for i in range(10):
    date+= pd.Timedelta(days=1)
    w_pd(date, content)
```
